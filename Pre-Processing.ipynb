```python
from statsmodels.tsa.stattools import adfuller
def adf_test(series):
    result = adfuller(series,autolag = 'AIC')
    print(f'ADF statistics: {result[0]}')
    print(f'p-value: {result[1]}')
    print('Critical Values')
    for key,values in result[4].items():
        print(f' {key} :{values}')
    if result[1]<=0.05:
        print("\nResult: Reject the null Hypothesis, Data is Non-Stationary.")
    else:
        print("\n Result: Fail to Reject the Null Hypothesis. Data is Non-Stationary")
print("Testing stationary data for the total Power:")
adf_test(df[target_col])
        
```

```python
# 1. Z-Score Method
df['z_score'] = np.abs(stats.zscore(df[target_col]))
z_outliers = df[df['z_score'] > 3]



# 2. IQR Method
Q1 = df[target_col].quantile(0.25)
Q3 = df[target_col].quantile(0.75)
IQR = Q3 - Q1
iqr_outliers = df[(df[target_col] < (Q1 - 1.5 * IQR)) | (df[target_col] > (Q3 + 1.5 * IQR))]

plt.figure(figsize=(15, 6))
plt.plot(df.index, df[target_col], label='Power Consumption', color='blue', alpha=0.5)
plt.scatter(z_outliers.index, z_outliers[target_col], color='red', label='Z-Score Outliers (|Z|>3)', zorder=5)
plt.scatter(iqr_outliers.index, iqr_outliers[target_col], marker='x', color='green', label='IQR Outliers', zorder=4)

plt.title('Mathematical Anomaly Detection in Energy Consumption')
plt.legend()
plt.show()

print(f"Number of Z-score outliers: {len(z_outliers)}")
print(f"Number of IQR outliers: {len(iqr_outliers)}")
```

```python
def calculate_entropy(series, bins=30):
    # Create a histogram of the data
    counts, _ = np.histogram(series, bins=bins, density=True)
    # Convert to probabilities
    probs = counts / np.sum(counts)
    # Remove zeros for log calculation
    probs = probs[probs > 0]
    # Calculate Shannon Entropy
    return -np.sum(probs * np.log2(probs))

entropy_val = calculate_entropy(df[target_col])
print(f"Shannon Entropy of {target_col}: {entropy_val:.4f} bits")

# Calculate entropy over a rolling window to see complexity changes over time
df['rolling_entropy'] = df[target_col].rolling(window=168).apply(calculate_entropy) # Weekly rolling

plt.figure(figsize=(12, 5))
df['rolling_entropy'].plot(color='purple')
plt.title('Rolling Shannon Entropy (System Complexity Over Time)')
plt.ylabel('Entropy (Bits)')
plt.show()
```

```python
def detect_cusum(x, threshold=5, drift=0.5):
    S_pos = np.zeros(len(x))
    S_neg = np.zeros(len(x))
    mu = np.mean(x)
    
    for i in range(1, len(x)):
        S_pos[i] = max(0, S_pos[i-1] + (x[i] - mu) - drift)
        S_neg[i] = max(0, S_neg[i-1] - (x[i] - mu) - drift)
    
    return S_pos, S_neg

S_pos, S_neg = detect_cusum(df[target_col].values)

plt.figure(figsize=(12, 6))
plt.plot(df.index, S_pos, label='CUSUM Positive (Upward Shift)', color='red')
plt.plot(df.index, S_neg, label='CUSUM Negative (Downward Shift)', color='blue')
plt.axhline(50, color='black', linestyle='--', label='Threshold')
plt.title('CUSUM Chart for Change Point Detection')
plt.legend()
plt.show()
```

```python
from numpy import polyfit
def hurts(ts):
    lags = range(2,100)
    tau =[np.sqrt(np.std(np.subtract(ts[lag:],ts[:-lag])))for lag in lags]
    poly = np.polyfit(np.log(lags),np.log(tau),1)
    return poly[0]*2.0

h_exp = hurts(df[target_col].dropna().values)
# h_exp = hurts(ts)
print(f"Hurst Expoenent of {target_col}: {h_exp:.4f}")
if h_exp>0.5:
    print("The series is persistent(Trending)")
else:
    print("The series is Anti-Persistent(Mean-Reverting)")
```

```python
# Discretize data into 3 states
bins = [0, df[target_col].quantile(0.33), df[target_col].quantile(0.66), np.inf]
labels = ['Low', 'Med', 'High']
df['state'] = pd.cut(df[target_col], bins=bins, labels=labels)

# Create transition matrix
state_transitions = pd.DataFrame({'current': df['state'], 'next': df['state'].shift(-1)})
transition_matrix = pd.crosstab(state_transitions['current'], state_transitions['next'], normalize='index')

plt.figure(figsize=(8, 6))
sns.heatmap(transition_matrix, annot=True, cmap='YlGnBu')
plt.title('Markov State Transition Probability Matrix')
plt.show()
```
