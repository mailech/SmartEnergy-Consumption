# Feature Engineering for Energy Consumption

This notebook creates advanced features for time-series forecasting.

```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/energy_data.csv")
df['timestamp'] = pd.to_datetime(df['timestamp'])
df.set_index('timestamp', inplace=True)
```

```python
df['lag_1'] = df['total_power'].shift(1)
df['lag_24'] = df['total_power'].shift(24)
```

```python
df['rolling_mean_24'] = df['total_power'].rolling(window=24).mean()
```

```python
df.dropna(inplace=True)
```
