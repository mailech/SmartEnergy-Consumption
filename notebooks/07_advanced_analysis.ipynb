# Advanced Energy Consumption Analysis

This notebook performs advanced analysis including volatility and device correlation to understand energy usage patterns.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/raw/energy_data.csv")

df['timestamp'] = pd.to_datetime(df['timestamp'])
df.set_index('timestamp', inplace=True)

df.head()
```

```python
df['volatility'] = df['total_power'].rolling(window=24).std()

df['volatility'].plot(figsize=(10,4), title="24 Hour Energy Volatility")
plt.show()
```

```python
cols = ['fridge','ac','lights','microwave']

sns.heatmap(df[cols].corr(), annot=True, cmap='coolwarm')
plt.title("Device-Level Correlation")
plt.show()
```

## Observations

- Volatility shows how energy usage fluctuates over time.
- Higher volatility indicates unstable energy consumption.
- Correlation heatmap shows relationships between devices.
- Helps identify which appliances contribute more to energy usage patterns.
